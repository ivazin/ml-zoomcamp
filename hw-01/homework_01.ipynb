## Q6. Median Value of Horsepower

1. Find the median value of the `horsepower` column.
2. Calculate the most frequent value (mode) of `horsepower`.
3. Use `fillna` to fill missing values with the mode.
4. Calculate the median value of `horsepower` again.
5. Has it changed?

```python
median_before = df["horsepower"].median()
mode_hp = df["horsepower"].mode()[0]
hp_filled = df["horsepower"].fillna(mode_hp)
median_after = hp_filled.median()

print(f"Median before: {median_before}")
print(f"Most frequent value (mode): {mode_hp}")
print(f"Median after fillna: {median_after}")

if median_after < median_before:
    print("Has it changed? -> Yes, it decreased")
elif median_after > median_before:
    print("Has it changed? -> Yes, it increased")
else:
    print("Has it changed? -> No")
```

## Q7. Sum of Weights (Linear Regression via Normal Equation)

1. Select all the cars from Asia.
2. Select only columns `vehicle_weight` and `model_year`.
3. Select the first 7 values.
4. Get the underlying NumPy array $X$.
5. Compute matrix-matrix multiplication $X^T X$.
6. Invert $X^T X$.
7. Create an array $y = [1100, 1300, 800, 900, 1000, 1100, 1200]$.
8. Multiply $(X^T X)^{-1}$ with $X^T$, then with $y$ to get $w$.
9. What is the sum of all elements of $w$?

```python
asia_subset = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].iloc[:7]
X = asia_subset.to_numpy()
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)

print("Resulting weight vector w:", w)
print(f"Sum of elements of w: {w.sum():.4f}")
```
